### DATA READING

```python
p_file_name=dbutils.widgets.get("file_name")
```

```python
df= spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format", "parquet")\
    .option("cloudFiles.schemaLocation",f"abfss://bronze@dbenttoend.dfs.core.windows.net/Checkpoint_{p_file_name}")\
        .load(f"abfss://source@dbenttoend.dfs.core.windows.net/{p_file_name}")
```

#### DATA WRITING

```python
df.writeStream.format("parquet")\
    .option("checkpointLocation",f"abfss://bronze@dbenttoend.dfs.core.windows.net/Checkpoint_{p_file_name}")\
    .option("path",f"abfss://bronze@dbenttoend.dfs.core.windows.net/{p_file_name}")\
    .trigger(once=True)\
    .start()
```
